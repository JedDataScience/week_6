# NBA Player Scouting Report for Mr. Rooney
## Using Data Science to Find the Right Players for the Wizards

## Overview

Mr. Rooney, this report uses advanced data analysis to group NBA players by their playing style not by their reputation or salary. By finding patterns in the numbers, we identified which players fit our team needs and which ones are overpriced for what they actually do on the court.

## What We Did


### Step 1: Gathered the Data
We collected every NBA player's statistics for the 2024-25 season—points scored, rebounds, minutes played, field goals, free throws, steals, blocks, etc.and then matched that with their 2025-26 salary information. We then cleaned the data selecting features that would be good predictors of salary and performance. 

### Step 2: Let the Computer Find Player Types  
Using a technique called "clustering," we let the computer group players based purely on how they play basketball. The computer doesn't know who's famous or who's highly paid it will just look at the stats and will group similar players together. For example:
- High scorers with lots of minutes tend to cluster together
- Rebounders and defenders cluster together
- Bench players with limited minutes form their own group
- ect...

### Step 3: Found the Right Number of Groups
We tested different numbers of player groups 5 and 2. We started with **5 groups** because that captured the different player types well. Then we used two scientific methods to double-check:

1. **The Elbow Method:** This shows us when adding more groups stops being helpful (like when diminishing returns start)
2. **Silhouette Score:** This measures how well players fit into their assigned groups


### Step 4: Visualized the Clusters
After grouping players, we created charts showing:
- How players cluster based on Points and Rebounds
- How salary overlays on top of these performance clusters

This visual approach reveals the hidden gems—great players on cheap contracts.

## How We Chose the Right Number of Groups

Before making any player recommendations, we needed to determine how many groups (clusters) make sense. Too few groups oversimplifies things and too many creates confusion. We used two standard methods:

- **The Elbow Method**
This shows when adding more groups stops giving us better insights. The graph below plots "within-cluster variation" (basically, how tight the groups are). The "elbow" is where the line bends—that's the sweet spot.
- **Silhouette Score**
This measures how well players fit into their assigned groups. Higher scores mean better-defined clusters.
